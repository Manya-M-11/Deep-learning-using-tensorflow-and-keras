```python
import math

def sigmoid(x):
    return 1/(1+math.exp(-x))

```

```python
sigmoid(100)
```

```python
sigmoid(-1)
```

```python
sigmoid(1)
```

```python
def tanh(x):
    return (math.exp(x) - math.exp(-x))/(math.exp(x)+ math.exp(-x))
```

```python
tanh(1)
```

```python
tanh(-1)
```

```python
tanh(50)
```

```python
tanh(-56)
```

```python
def Relu(x):
    return max(0,x)
```

```python
Relu(1)
```

```python
Relu(100)
```

```python
Relu(-9)
```

```python
def leaky(x):
    return max(0.1*x,x)
```

```python
leaky(100)
```

```python
leaky(-8)
```
